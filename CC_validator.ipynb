### Credit Card Validator

This validator will check if a given credit card number is valid based on:
1.  **Luhn Algorithm**: A simple checksum formula used to validate a variety of identification numbers, such as credit card numbers, IMEI numbers, etc.
2.  **Card Company Patterns**: Specific prefixes and lengths for popular credit card companies like Visa, Mastercard, and American Express.

```python
def luhn_checksum(card_number):
    digits = [int(d) for d in card_number if d.isdigit()]
    if not digits:
        return False

    total_sum = 0
    num_digits = len(digits)
    is_second = False

    for i in range(num_digits - 1, -1, -1):
        digit = digits[i]
        if is_second:
            digit *= 2
            if digit > 9:
                digit = digit - 9
        total_sum += digit
        is_second = not is_second

    return total_sum % 10 == 0

def validate_credit_card(card_number):
    card_number = card_number.replace(' ', '').replace('-', '')

    if not card_number.isdigit():
        return "Invalid: Contains non-digit characters."

    if not luhn_checksum(card_number):
        return "Invalid: Fails Luhn algorithm checksum."

    length = len(card_number)
    first_digit = card_number[0]
    first_two_digits = card_number[0:2]
    first_four_digits = card_number[0:4]

    # Visa
    if first_digit == '4' and (length == 13 or length == 16):
        return "Valid: Visa"
    # Mastercard
    elif first_two_digits in ['51', '52', '53', '54', '55'] and length == 16:
        return "Valid: Mastercard"
    elif 2221 <= int(first_four_digits) <= 2720 and length == 16:
        return "Valid: Mastercard"
    # American Express
    elif first_two_digits in ['34', '37'] and length == 15:
        return "Valid: American Express"
    # Discover (often starts with 6011, 644-649, 65, length 16 or 19)
    elif first_four_digits == '6011' and length == 16:
        return "Valid: Discover"
    elif first_two_digits == '65' and length == 16:
        return "Valid: Discover"
    elif first_three_digits == '644' and length == 16:
        return "Valid: Discover"
    # Diners Club/Carte Blanche (starts with 300-305, 309, 36, 38, length 14)
    elif first_two_digits in ['36', '38'] and length == 14:
        return "Valid: Diners Club/Carte Blanche"
    elif first_three_digits in ['300', '301', '302', '303', '304', '305', '309'] and length == 14:
        return "Valid: Diners Club/Carte Blanche"
    # JCB (starts with 3528-3589, length 16)
    elif 3528 <= int(first_four_digits) <= 3589 and length == 16:
        return "Valid: JCB"
    else:
        return "Valid: Card company not identified or pattern doesn't match a known type."

# Get input from the user
card_number_input = input("Enter the credit card number: ")

# Validate and print the result
result = validate_credit_card(card_number_input)
print(result)
```
