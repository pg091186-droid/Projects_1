### Factorial Calculation

This notebook will demonstrate two ways to calculate the factorial of a positive integer: iteratively and recursively. You will be prompted to enter a positive integer and then choose which method you'd like to use.

```python
# Get a positive integer from the user
while True:
    try:
        num = int(input("Enter a positive integer: "))
        if num < 0:
            print("Please enter a positive integer.")
        else:
            break
    except ValueError:
        print("Invalid input. Please enter an integer.")

print(f"You entered: {num}")
```

```python
def factorial_iterative(n):
    """Calculates the factorial of a non-negative integer using iteration."""
    if n == 0:
        return 1
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

def factorial_recursive(n):
    """Calculates the factorial of a non-negative integer using recursion."""
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial_recursive(n - 1)

print("Factorial functions (iterative and recursive) are defined.")
```
