This program allows you to calculate GST and total amounts in two ways:

1.  **Calculate Total Amount from Base Cost**: Provide a base cost and a GST category (e.g., A, B, C), and it will compute the GST amount and the final total amount.
2.  **Calculate Base Cost from Total Amount**: Provide a total amount and the applicable GST percentage, and it will compute the base cost and the GST amount.

```python
# Define GST categories and their percentages
gst_categories = {
    'A': 5,   # 5% GST
    'B': 12,  # 12% GST
    'C': 18,  # 18% GST
    'D': 28   # 28% GST
}

def calculate_total_from_base(base_cost, gst_percentage):
    """
    Calculates GST amount and total amount from a given base cost and GST percentage.
    """
    if not isinstance(base_cost, (int, float)) or base_cost < 0:
        raise ValueError("Base cost must be a non-negative number.")
    if not isinstance(gst_percentage, (int, float)) or not (0 <= gst_percentage <= 100):
        raise ValueError("GST percentage must be between 0 and 100.")

    gst_amount = base_cost * (gst_percentage / 100)
    total_amount = base_cost + gst_amount
    return gst_amount, total_amount

def calculate_base_from_total(total_amount, gst_percentage):
    """
    Calculates base cost and GST amount from a given total amount and GST percentage.
    """
    if not isinstance(total_amount, (int, float)) or total_amount < 0:
        raise ValueError("Total amount must be a non-negative number.")
    if not isinstance(gst_percentage, (int, float)) or not (0 <= gst_percentage <= 100):
        raise ValueError("GST percentage must be between 0 and 100.")

    # total_amount = base_cost * (1 + gst_percentage / 100)
    # base_cost = total_amount / (1 + gst_percentage / 100)
    base_cost = total_amount / (1 + gst_percentage / 100)
    gst_amount = total_amount - base_cost
    return base_cost, gst_amount

# Main program logic
print("Choose an option:")
print("1. Calculate Total Amount from Base Cost")
print("2. Calculate Base Cost from Total Amount")

while True:
    try:
        choice = input("Enter your choice (1 or 2): ")
        if choice == '1':
            base_cost = float(input("Enter the base cost: "))
            print("Available GST Categories:")
            for category, percentage in gst_categories.items():
                print(f"  {category}: {percentage}%")
            while True:
                gst_category_input = input("Enter the GST category (A, B, C, D): ").upper()
                if gst_category_input in gst_categories:
                    gst_percentage = gst_categories[gst_category_input]
                    break
                else:
                    print("Invalid GST category. Please choose from A, B, C, D.")

            gst_amount, total_amount = calculate_total_from_base(base_cost, gst_percentage)
            print(f"\n--- Calculation Results ---")
            print(f"Base Cost: ${base_cost:.2f}")
            print(f"GST Percentage: {gst_percentage}%")
            print(f"GST Amount: ${gst_amount:.2f}")
            print(f"Total Amount: ${total_amount:.2f}")
            break

        elif choice == '2':
            total_amount = float(input("Enter the total amount: "))
            gst_percentage = float(input("Enter the applicable GST percentage (e.g., 5 for 5%): "))

            base_cost, gst_amount = calculate_base_from_total(total_amount, gst_percentage)
            print(f"\n--- Calculation Results ---")
            print(f"Total Amount: ${total_amount:.2f}")
            print(f"GST Percentage: {gst_percentage}%")
            print(f"Base Cost: ${base_cost:.2f}")
            print(f"GST Amount: ${gst_amount:.2f}")
            break

        else:
            print("Invalid choice. Please enter 1 or 2.")
    except ValueError as e:
        print(f"Input Error: {e}. Please enter valid numbers.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```
