```python
from decimal import Decimal, getcontext

def calculate_e(n):
    # Set precision slightly higher to avoid rounding errors
    getcontext().prec = n + 5

    e = Decimal(1)
    factorial = Decimal(1)
    i = 1

    while True:
        factorial *= i
        term = Decimal(1) / factorial

        if term < Decimal(10) ** (-(n + 2)):
            break

        e += term
        i += 1

    return +e  # Apply precision

# -------- Main Program --------
n = int(input("Enter number of decimal places: "))

e_value = calculate_e(n)

# Print exactly n decimal places
print(f"Value of e up to {n} decimal places:")
print(str(e_value)[:n+2])   # +2 for "2."
```
