```python
# Function to check if a number is prime
def is_prime(n):
    # Prime numbers are natural numbers greater than 1
    # 1 is not a prime number
    if n <= 1:
        return False
    # 2 and 3 are prime numbers
    if n <= 3:
        return True
    # All primes greater than 3 are of the form 6k ± 1
    # This optimization skips checking divisibility by multiples of 2 and 3
    if n % 2 == 0 or n % 3 == 0:
        return False
    # Check for divisibility from 5 up to sqrt(n)
    # We increment by 6 because we've already handled multiples of 2 and 3.
    # So, we only need to check numbers of the form 6k-1 and 6k+1.
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True

# Function to find the next prime number greater than a given number
def find_next_prime(current_prime):
    # If current_prime is less than 2, the next prime is 2
    if current_prime < 2:
        return 2
    # If current_prime is 2, the next prime is 3
    if current_prime == 2:
        return 3

    # Start checking from the number after current_prime
    num = current_prime + 1

    # If num is even, increment by 1 to make it odd (all primes > 2 are odd)
    if num % 2 == 0:
        num += 1

    while True:
        if is_prime(num):
            return num
        num += 2 # Increment by 2 to only check odd numbers

# Main program loop for user interaction
print("Let's find prime numbers!")
current_prime = 1 # Start checking from the number after 1

while True:
    current_prime = find_next_prime(current_prime)
    print(f"The next prime number is: {current_prime}")

    user_input = input("Do you want to find the next prime? (yes/no): ").lower()
    if user_input != 'yes':
        print("Stopping prime number generation. Goodbye!")
        break
```
