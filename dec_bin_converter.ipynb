### Decimal and Binary Converter

This program includes functions to convert a decimal (base-10) number to its binary (base-2) representation and vice-versa.

```python
def decimal_to_binary(decimal_num):
    if decimal_num == 0:
        return "0"

    binary_num = ""
    while decimal_num > 0:
        remainder = decimal_num % 2
        binary_num = str(remainder) + binary_num
        decimal_num = decimal_num // 2
    return binary_num

def binary_to_decimal(binary_string):
    decimal_num = 0
    power = 0
    # Iterate through the binary string from right to left
    for digit in reversed(binary_string):
        if digit == '1':
            decimal_num += 2**power
        elif digit != '0':
            raise ValueError("Invalid binary input: contains non-binary digits")
        power += 1
    return decimal_num

# Example Usage: Decimal to Binary
decimal_input_1 = 25
binary_output_1 = decimal_to_binary(decimal_input_1)
print(f"The decimal number {decimal_input_1} is {binary_output_1} in binary.")

decimal_input_2 = 10
binary_output_2 = decimal_to_binary(decimal_input_2)
print(f"The decimal number {decimal_input_2} is {binary_output_2} in binary.")

print("\n---\n")

# Example Usage: Binary to Decimal
binary_input_1 = "11001"
decimal_output_1 = binary_to_decimal(binary_input_1)
print(f"The binary number {binary_input_1} is {decimal_output_1} in decimal.")

binary_input_2 = "1010"
decimal_output_2 = binary_to_decimal(binary_input_2)
print(f"The binary number {binary_input_2} is {decimal_output_2} in decimal.")
```
