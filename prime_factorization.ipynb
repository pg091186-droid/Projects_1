```python
import math

# Step 1: Take input from the user
N = int(input("Enter a number to factorize: "))

# Step 2: Handle edge case
if N < 2:
    print("No prime factors for numbers less than 2.")
    exit()

# Step 3: Generate list of primes up to √N using Sieve of Eratosthenes
limit = int(math.isqrt(N)) + 1
is_prime = [True] * (limit + 1)
is_prime[0:2] = [False, False]

for i in range(2, int(math.isqrt(limit)) + 1):
    if is_prime[i]:
        for j in range(i*i, limit + 1, i):
            is_prime[j] = False

primes = [i for i, val in enumerate(is_prime) if val]

# Step 4: Initialize list to store prime factors
prime_factors = []

# Step 5: Loop through primes and divide N
for p in primes:
    while N % p == 0:
        prime_factors.append(p)
        N //= p

# Step 6: If N is still greater than 1, it is prime
if N > 1:
    prime_factors.append(N)

# Step 7: Display the result
print("Prime factors:", prime_factors)
```
