```python
# Basic Calculator Program

# Take input from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
operation = input(
    "Enter the operation (+, -, *, /, **, %): "
)

# Perform calculation
if operation == "+":
    result = num1 + num2

elif operation == "-":
    result = num1 - num2

elif operation == "*":
    result = num1 * num2

elif operation == "/":
    if num2 == 0:
        print("Error: Division by zero is not allowed.")
        exit()
    result = num1 / num2

elif operation == "**":
    result = num1 ** num2

elif operation == "%":
    result = num1 % num2

else:
    print("Invalid operation.")
    exit()

# Display result
print("Result:", result)
```
