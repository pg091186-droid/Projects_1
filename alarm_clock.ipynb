```python
import datetime
import time
import pytz # Import pytz for timezone handling

def set_alarm():
    print("Alarm Clock")
    print("1. Set alarm after X minutes")
    print("2. Set alarm at a specific time (HH:MM)")

    # Define IST timezone
    IST = pytz.timezone('Asia/Kolkata')

    alarm_time = None

    while alarm_time is None: # Loop until a valid alarm_time is set
        choice = input("Enter your choice (1 or 2): ")

        if choice == '1':
            while True: # Loop for valid minutes input
                try:
                    minutes_str = input("Enter the number of minutes: ")
                    if '.' in minutes_str:
                        print("Invalid input. Please enter a whole number for minutes.")
                        continue
                    minutes = int(minutes_str)
                    # Get current time in IST
                    now_ist = datetime.datetime.now(IST)
                    alarm_time = now_ist + datetime.timedelta(minutes=minutes)
                    print(f"Alarm set for {minutes} minutes from now. It will ring at {alarm_time.strftime('%H:%M:%S %Z%z')}")
                    break
                except ValueError:
                    print("Invalid input. Please enter a whole number for minutes.")
        elif choice == '2':
            while True: # Loop for valid time input
                time_str = input("Enter the alarm time in HH:MM format: ")
                try:
                    hour, minute = map(int, time_str.split(':'))
                    # Get current time in IST
                    now_ist = datetime.datetime.now(IST)
                    # Set alarm time with today's date in IST
                    alarm_time = now_ist.replace(hour=hour, minute=minute, second=0, microsecond=0)

                    if alarm_time < now_ist:
                        alarm_time += datetime.timedelta(days=1)

                    print(f"Alarm set for {alarm_time.strftime('%H:%M:%S %Z%z')}.")
                    break
                except ValueError:
                    print("Invalid time format. Please use HH:MM.")
        else:
            print("Invalid choice. Please enter 1 or 2.")

    print("Waiting for alarm...")
    while True:
        now_ist = datetime.datetime.now(IST)
        time_until_alarm = (alarm_time - now_ist).total_seconds()

        if time_until_alarm <= 0:
            print("\n=====================================")
            print("          ALARM! ALARM! ALARM!       ")
            print("=====================================")
            break
        else:
            if time_until_alarm > 5:
                sleep_duration = min(30, time_until_alarm - 2)
                time.sleep(sleep_duration)
            else:
                time.sleep(1)

set_alarm()
```
