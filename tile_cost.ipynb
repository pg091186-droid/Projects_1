```python
# Get the width of the floor from the user
# Use float() to allow for decimal values in width
width = float(input("Enter the width of the floor (e.g., in meters or feet): "))

# Get the height (or length) of the floor from the user
# Use float() to allow for decimal values in height
height = float(input("Enter the height/length of the floor (e.g., in meters or feet): "))

# Get the cost per unit area of the tile from the user
# Use float() to allow for decimal values in cost
cost_per_unit_area = float(input("Enter the cost per unit area of the tile (e.g., $ per sq meter/foot): "))

# Calculate the total area of the floor
# Area = Width * Height
total_area = width * height

# Calculate the total cost of the tiles
# Total Cost = Total Area * Cost per Unit Area
total_cost = total_area * cost_per_unit_area

# Display the calculated total area and total cost to the user
print(f"\nFloor Dimensions: {width} x {height}")
print(f"Total Floor Area: {total_area:.2f} square units")
print(f"Cost per unit area: ${cost_per_unit_area:.2f}")
print(f"Total Cost to tile the floor: ${total_cost:.2f}")
```
