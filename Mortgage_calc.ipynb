# Task
Develop a mortgage calculator that can collect principal loan amount, annual interest rate, loan term, and compounding interval (Monthly, Weekly, Daily, or Continually). ## Get Mortgage Details

### Subtask:
Develop a function to collect all necessary mortgage details from the user, including the principal loan amount, annual interest rate, loan term in years, and the desired compounding interval (Monthly, Weekly, Daily, or Continually). Also, include an option to ask if the user intends to make a part payment.


**Reasoning**:
To begin developing the mortgage details collection function, I will define the `get_mortgage_details` function that prompts the user for all the required inputs, including the principal loan amount, annual interest rate, loan term, compounding interval, and part payment intention, with basic input validation.



```python
def get_mortgage_details():
    """Collects all necessary mortgage details from the user."""

    # 1. Principal Loan Amount
    while True:
        try:
            principal_amount_str = input("Enter the principal loan amount (e.g., 200000): ")
            principal_amount = float(principal_amount_str)
            if principal_amount <= 0:
                print("Principal loan amount must be a positive number. Please try again.")
            else:
                break
        except ValueError:
            print("Invalid input. Please enter a numerical value for the principal loan amount.")

    # 2. Annual Interest Rate
    while True:
        try:
            annual_interest_rate_str = input("Enter the annual interest rate (as a percentage, e.g., 5.0 for 5%): ")
            annual_interest_rate = float(annual_interest_rate_str)
            if annual_interest_rate <= 0:
                print("Annual interest rate must be a positive number. Please try again.")
            else:
                # Convert to decimal for calculations later
                annual_interest_rate_decimal = annual_interest_rate / 100
                break
        except ValueError:
            print("Invalid input. Please enter a numerical value for the annual interest rate.")

    # 3. Loan Term in Years
    while True:
        try:
            loan_term_years_str = input("Enter the loan term in years (e.g., 30): ")
            loan_term_years = int(loan_term_years_str)
            if loan_term_years <= 0:
                print("Loan term must be a positive integer. Please try again.")
            else:
                break
        except ValueError:
            print("Invalid input. Please enter an integer value for the loan term.")

    # 4. Compounding Interval
    valid_intervals = ['monthly', 'weekly', 'daily', 'continually']
    while True:
        compounding_interval = input(f"Select the compounding interval ({', '.join(valid_intervals)}): ").lower()
        if compounding_interval in valid_intervals:
            break
        else:
            print("Invalid compounding interval. Please choose from 'Monthly', 'Weekly', 'Daily', or 'Continually'.")

    # 5. Part Payment Option
    while True:
        make_part_payment = input("Do you intend to make a part payment? (yes/no): ").lower()
        if make_part_payment in ['yes', 'no']:
            break
        else:
            print("Invalid input. Please answer 'yes' or 'no'.")

    return principal_amount, annual_interest_rate_decimal, loan_term_years, compounding_interval, make_part_payment

print("Function 'get_mortgage_details' defined successfully.")
```

## Calculate Initial Monthly Payment and Loan Duration

### Subtask:
Implement the formulas to calculate the effective monthly interest rate and then the monthly payment for a fixed-term mortgage, considering various compounding intervals. Based on this, calculate the total number of payments and the total loan duration.


**Reasoning**:
I will define the `calculate_mortgage_details` function as instructed, implementing the formulas for effective interest rates, total payments, and monthly payment, considering different compounding intervals, including a special case for continuous compounding. I'll also import the `math` module as required.



```python
import math

def calculate_mortgage_details(principal_amount, annual_interest_rate_decimal, loan_term_years, compounding_interval):
    """Calculates the effective monthly interest rate, total payments, and monthly payment."""

    # 2. Determine compounding frequency per year (n)
    if compounding_interval == 'monthly':
        n = 12
    elif compounding_interval == 'weekly':
        n = 52
    elif compounding_interval == 'daily':
        n = 365
    elif compounding_interval == 'continually':
        n = None # Handled separately for continuous compounding
    else:
        raise ValueError("Invalid compounding interval provided.")

    # 3. Calculate the Effective Annual Rate (EAR)
    if compounding_interval == 'continually':
        ear = math.exp(annual_interest_rate_decimal) - 1
    else:
        ear = (1 + annual_interest_rate_decimal / n)**n - 1

    # 4. Calculate the Effective Monthly Rate (EMR)
    emr = (1 + ear)**(1/12) - 1

    # 5. Calculate the total number of monthly payments (N)
    N = loan_term_years * 12

    # 6. Calculate the monthly payment
    if emr == 0:
        monthly_payment = principal_amount / N
    else:
        monthly_payment = principal_amount * (emr * (1 + emr)**N) / ((1 + emr)**N - 1)

    return monthly_payment, N

print("Function 'calculate_mortgage_details' defined successfully.")
```

**Reasoning**:
Now that the `calculate_mortgage_details` function is defined, I will call it with the previously collected mortgage details and then print the calculated monthly payment and total loan duration, as per the subtask instructions.



```python
monthly_payment, total_payments = calculate_mortgage_details(
    principal_amount, annual_interest_rate_decimal, loan_term_years, compounding_interval
)

total_loan_duration_years = total_payments / 12

print(
    f"\n--- Calculated Mortgage Details ---"
)
print(f"Initial Monthly Payment: ${monthly_payment:,.2f}")
print(f"Total Number of Payments: {int(total_payments)}")
print(f"Total Loan Duration: {total_loan_duration_years:.2f} years")
```

## Handle Part Payment and Recalculation

### Subtask:
If the user chooses to make a part payment, develop logic to accept the part payment amount, recalculate the remaining principal, and then adjust either the monthly payment or the remaining loan duration based on the user's preference or a predefined strategy (e.g., keeping the monthly payment same and reducing term).


**Reasoning**:
I will check if the user intended to make a part payment. If so, I will prompt for the part payment amount with validation, calculate the remaining principal, and then ask the user to choose between reducing the monthly payment or the loan term. This involves calculating the effective monthly rate and applying the relevant mortgage formulas to adjust the payment or duration.



```python
if make_part_payment == 'yes':
    print("\n--- Part Payment Details ---")
    while True:
        try:
            part_payment_amount_str = input(f"Enter the part payment amount (Current Principal: ${principal_amount:,.2f}): ")
            part_payment_amount = float(part_payment_amount_str)
            if part_payment_amount <= 0 or part_payment_amount > principal_amount:
                print("Part payment must be a positive number and not exceed the current principal amount. Please try again.")
            else:
                break
        except ValueError:
            print("Invalid input. Please enter a numerical value for the part payment amount.")

    remaining_principal = principal_amount - part_payment_amount
    print(f"Part payment of ${part_payment_amount:,.2f} applied. Remaining Principal: ${remaining_principal:,.2f}")

    # Determine EMR (reusing logic from calculate_mortgage_details)
    if compounding_interval == 'monthly':
        n_compounding = 12
    elif compounding_interval == 'weekly':
        n_compounding = 52
    elif compounding_interval == 'daily':
        n_compounding = 365
    elif compounding_interval == 'continually':
        n_compounding = None # Handled separately for continuous compounding

    if compounding_interval == 'continually':
        ear = math.exp(annual_interest_rate_decimal) - 1
    else:
        ear = (1 + annual_interest_rate_decimal / n_compounding)**n_compounding - 1
    emr = (1 + ear)**(1/12) - 1

    valid_part_payment_options = ['reduce_monthly_payment', 'reduce_loan_term']
    while True:
        part_payment_option = input(f"Do you want to '{valid_part_payment_options[0]}' or '{valid_part_payment_options[1]}'? ").lower()
        if part_payment_option in valid_part_payment_options:
            break
        else:
            print("Invalid option. Please choose 'reduce_monthly_payment' or 'reduce_loan_term'.")

    if part_payment_option == 'reduce_loan_term':
        # Recalculate total_payments (N) with remaining_principal and original monthly_payment
        if emr == 0:
            new_total_payments = remaining_principal / monthly_payment
        else:
            # P = PMT * [1 - (1 + EMR)^-N] / EMR
            # (P * EMR) / PMT = 1 - (1 + EMR)^-N
            # (1 + EMR)^-N = 1 - (P * EMR) / PMT
            # -N * log(1 + EMR) = log(1 - (P * EMR) / PMT)
            # N = -log(1 - (P * EMR) / PMT) / log(1 + EMR)
            term_numerator = 1 - (remaining_principal * emr) / monthly_payment
            if term_numerator <= 0: # Handle cases where monthly_payment is too low to pay off principal with interest
                print("Error: Monthly payment is too low to cover remaining principal with interest. Cannot reduce loan term.")
                new_total_payments = total_payments # Revert to original
            else:
                new_total_payments = -math.log(term_numerator) / math.log(1 + emr)

        new_total_loan_duration_years = new_total_payments / 12
        print(f"\n--- After Part Payment (Reduced Loan Term) ---")
        print(f"Remaining Principal: ${remaining_principal:,.2f}")
        print(f"Original Monthly Payment: ${monthly_payment:,.2f}")
        print(f"New Total Number of Payments: {int(new_total_payments)}")
        print(f"New Total Loan Duration: {new_total_loan_duration_years:.2f} years")
        # Update global variables for potential future use
        principal_amount = remaining_principal
        total_payments = new_total_payments
        total_loan_duration_years = new_total_loan_duration_years

    elif part_payment_option == 'reduce_monthly_payment':
        # Recalculate monthly_payment with remaining_principal and original total_payments
        if emr == 0:
            new_monthly_payment = remaining_principal / total_payments
        else:
            new_monthly_payment = remaining_principal * (emr * (1 + emr)**total_payments) / ((1 + emr)**total_payments - 1)

        print(f"\n--- After Part Payment (Reduced Monthly Payment) ---")
        print(f"Remaining Principal: ${remaining_principal:,.2f}")
        print(f"Original Loan Duration: {total_loan_duration_years:.2f} years ({int(total_payments)} payments)")
        print(f"New Monthly Payment: ${new_monthly_payment:,.2f}")
        # Update global variables for potential future use
        principal_amount = remaining_principal
        monthly_payment = new_monthly_payment

else:
    print("\nNo part payment intended.")
```
